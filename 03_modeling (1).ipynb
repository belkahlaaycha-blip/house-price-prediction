This notebook focuses on training multiple regression models to predict house prices.

The goal of this notebook is to train and compare several regression models
to predict house prices using the prepared dataset.

```python
# 03_Modeling
# Goal: Train ML models for predicting house prices
# -  baseline model (Linear Regression)
# -  improved model (Ridge Regression)
# - Optional advanced models: Lasso Regression, Polynomial Regression
# - Export models and predictions for Evaluation notebook
```

Imports machine learning models and tools needed for training models

Libraries used:

LinearRegression → basic regression model

Ridge → linear regression with regularization

Lasso → linear regression with feature selection

PolynomialFeatures → transforms features into polynomial form

Pipeline → combines multiple steps into one model

```python
# 1.Import required libraries
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score
import joblib
# 2. Load prepared training and testing datasets

X_train = pd.read_csv(r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\X_train.csv")
X_test  = pd.read_csv(r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\X_test.csv")
y_train = pd.read_csv(r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_train.csv")
y_test  = pd.read_csv(r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_test.csv")

print("Training and testing datasets loaded successfully.")
print("X_train shape:", X_train.shape, "X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape, "y_test shape:", y_test.shape)
```

What this cell does:

Loads cleaned and prepared data from the previous notebook

Ensures modeling is independent from data preparation

Data loaded:

x_train → input features for training

x_test → input features for testing

y_train → target values for training

y_test → target values for testing

What this cell does:

Creates a linear regression model

Trains it on training data

Generates predictions

```python
#3. Baseline Model: Linear Regression
# nitialize Linear Regression model
lr_model = LinearRegression()
# Train the model
lr_model.fit(X_train, y_train)

# Save model
joblib.dump(lr_model, r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\linear_model.pkl")

# Coefficients & intercept
print("Linear Regression coefficients:", lr_model.coef_)
print("Linear Regression intercept:", lr_model.intercept_)
```

```python
# Predictions
y_train_pred_lr = lr_model.predict(X_train)
y_test_pred_lr  = lr_model.predict(X_test)
```

Save Linear Regression predictions for later evaluation

```python
# Save predictions
pd.DataFrame(y_train_pred_lr, columns=["y_train_pred_lr"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_train_pred_lr.csv", index=False)

pd.DataFrame(y_test_pred_lr, columns=["y_test_pred_lr"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_test_pred_lr.csv", index=False)

print("Linear Regression training complete. Predictions saved.")
```

What this cell does (for your understanding)

Initializes a Ridge Regression model with regularization (alpha=1.0)

Trains the model using the training data

Saves the trained model as a .pkl file using joblib

Ensures reproducibility and notebook independence

```python
# 4️. Improved Model: Ridge Regression (L2 Regularization)
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_train, y_train)

# Save model
joblib.dump(ridge_model, r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\ridge_model.pkl")
```

```python
# Predictions
y_train_pred_ridge = ridge_model.predict(X_train)
y_test_pred_ridge  = ridge_model.predict(X_test)

# Save predictions
pd.DataFrame(y_train_pred_ridge, columns=["y_train_pred_ridge"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_train_pred_ridge.csv", index=False)

pd.DataFrame(y_test_pred_ridge, columns=["y_test_pred_ridge"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_test_pred_ridge.csv", index=False)

print("Ridge Regression training complete. Predictions saved.")
```

Initializes a Lasso Regression model with L1 regularization

Trains the model using training data

Generates predictions for training and test datasets

Saves predictions to CSV files

Prepares outputs for the Evaluation notebook

```python
# 5️. Optional: Lasso Regression (L1 Regularization)

lasso_model = Lasso(alpha=0.01)
lasso_model.fit(X_train, y_train)

# Predictions
y_train_pred_lasso = lasso_model.predict(X_train)
y_test_pred_lasso  = lasso_model.predict(X_test)

# Save predictions
pd.DataFrame(y_train_pred_lasso, columns=["y_train_pred_lasso"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_train_pred_lasso.csv", index=False)

pd.DataFrame(y_test_pred_lasso, columns=["y_test_pred_lasso"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_test_pred_lasso.csv", index=False)

print("Lasso Regression (optional) predictions saved.")
```

Transforms the original features into polynomial features of degree 2

Allows the model to learn non-linear patterns

Trains a Linear Regression model on the expanded feature space

Uses the same transformation for training and test data to ensure consistency

```python
# 6️. Optional: Polynomial Regression (degree=2)
poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(X_train)
X_test_poly  = poly.transform(X_test)

poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)
```

Uses the trained Polynomial Regression model to make predictions

Produces predictions for both training and test datasets

Saves the predictions to external files for reproducibility

Keeps modeling and evaluation phases independent

```python
# Predictions
y_train_pred_poly = poly_model.predict(X_train_poly)
y_test_pred_poly  = poly_model.predict(X_test_poly)

# Save predictions
pd.DataFrame(y_train_pred_poly, columns=["y_train_pred_poly"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_train_pred_poly.csv", index=False)

pd.DataFrame(y_test_pred_poly, columns=["y_test_pred_poly"]).to_csv(
    r"C:\Users\bka20\OneDrive\Desktop\Project ML\Data\y_test_pred_poly.csv", index=False)

print("Polynomial Regression (optional) predictions saved.")
```

```python

# 7️. Quick evaluation (for reference)

print("R² scores on test set:")
print("Linear Regression:", r2_score(y_test, y_test_pred_lr))
print("Ridge Regression:", r2_score(y_test, y_test_pred_ridge))
print("Lasso Regression:", r2_score(y_test, y_test_pred_lasso))
print("Polynomial Regression:", r2_score(y_test, y_test_pred_poly))

print("Modeling phase completed. All models and predictions saved for Evaluation notebook.")
```
