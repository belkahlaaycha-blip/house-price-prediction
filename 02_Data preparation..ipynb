# 02 – Data Preparation

## Purpose of This Notebook

This notebook focuses on **preparing the dataset for the modeling phase** of the House Price Prediction project.  
At this stage, the data has already been cleaned and normalized in the previous notebook.

The main goal here is to transform the dataset into a **model-ready format**, while ensuring:
- clarity,
- reproducibility,
- and absence of data leakage.

---

## What Is Done in This Notebook

The following steps are performed:

- **Load the cleaned and normalized dataset**  
  Ensures consistency and avoids repeating preprocessing steps.

- **Separate features and target variable**  
  Clearly defines inputs (house characteristics) and output (house price).

- **Split the dataset into training and testing sets**  
  Allows objective evaluation of model performance on unseen data.

- **Export the prepared datasets**  
  Enables smooth transition to the modeling notebook and better project organization.

---

## Why This Step Is Important

Data preparation is a **critical step in any machine learning project**:
- Poor preparation leads to unreliable models
- Proper separation of data prevents overfitting
- Well-structured datasets make modeling easier and more interpretable

By the end of this notebook, the data will be **fully ready for model training and evaluation**.

## **5. Split the Dataset into Training and Testing Sets**

After preparing the features and target, we now **split the dataset** into **training** and **testing** sets.  
This step ensures that we can **train the model** on one set of data and **evaluate its performance** on unseen data.

### **Step 1: Split the Data**
We use an **80/20 split**:
- 80% for training  
- 20% for testing  

We also set `random_state=42` to ensure the **same split every time** for reproducibility.

```python
